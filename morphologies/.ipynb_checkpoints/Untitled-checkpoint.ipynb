```python
import pandas as pd


def add_section_index_by_type(df):
    """
    Add section_index using a separate counter for each section_type.

    A new section starts when:
        1) the point has no parent, prev_point_id == -1
        2) the parent is missing from the dataframe
        3) the section_type changes from parent to child
        4) the parent has more than one child

    The section_index counter is independent for each section_type.
    """

    df = df.copy()

    if not df["point_id"].is_unique:
        raise ValueError("point_id must be unique.")

    child_count = df["prev_point_id"].value_counts().to_dict()

    point_to_parent = dict(zip(df["point_id"], df["prev_point_id"]))
    point_to_type = dict(zip(df["point_id"], df["section_type"]))

    point_to_section_index = {}
    section_counter_by_type = {}

    def new_section_index(section_type):
        if section_type not in section_counter_by_type:
            section_counter_by_type[section_type] = -1

        section_counter_by_type[section_type] += 1
        return section_counter_by_type[section_type]

    def assign_section_index(point_id):
        if point_id in point_to_section_index:
            return point_to_section_index[point_id]

        section_type = point_to_type[point_id]
        prev_point_id = point_to_parent[point_id]

        # Root or missing parent
        if prev_point_id == -1 or prev_point_id not in point_to_type:
            section_index = new_section_index(section_type)
            point_to_section_index[point_id] = section_index
            return section_index

        parent_type = point_to_type[prev_point_id]
        parent_child_count = child_count.get(prev_point_id, 0)

        # Start a new section
        if section_type != parent_type or parent_child_count > 1:
            section_index = new_section_index(section_type)
            point_to_section_index[point_id] = section_index
            return section_index

        # Continue parent's section
        section_index = assign_section_index(prev_point_id)
        point_to_section_index[point_id] = section_index
        return section_index

    for point_id in df["point_id"]:
        assign_section_index(point_id)

    df["section_index"] = df["point_id"].map(point_to_section_index).astype(int)

    return df


def add_parent_section_info(df):
    """
    Add parent_section_type and parent_section_index.

    For each section, the parent section is determined from the prev_point_id
    of the first point in that section.

    If a section has no parent, parent_section_type = "" and
    parent_section_index = -1.
    """

    df = df.copy()

    if not df["point_id"].is_unique:
        raise ValueError("point_id must be unique.")

    required_columns = {
        "point_id",
        "section_type",
        "section_index",
        "prev_point_id",
    }

    missing = required_columns - set(df.columns)
    if missing:
        raise ValueError(f"Missing required columns: {missing}")

    # Lookup parent point information
    point_lookup = df.set_index("point_id")[["section_type", "section_index"]]

    # One row per section: first point of each section
    section_parents = (
        df.groupby(["section_type", "section_index"], sort=False)
        .first()
        .reset_index()
        [["section_type", "section_index", "prev_point_id"]]
    )

    # Map prev_point_id to the parent's section_type and section_index
    section_parents["parent_section_type"] = (
        section_parents["prev_point_id"]
        .map(point_lookup["section_type"])
        .fillna("")
    )

    section_parents["parent_section_index"] = (
        section_parents["prev_point_id"]
        .map(point_lookup["section_index"])
        .fillna(-1)
        .astype(int)
    )

    # Merge parent-section info back to all points
    df = df.merge(
        section_parents[
            [
                "section_type",
                "section_index",
                "parent_section_type",
                "parent_section_index",
            ]
        ],
        on=["section_type", "section_index"],
        how="left",
    )

    return df


def read_swc_to_dataframe(
    swc_path,
    section_type_to_label=None,
):
    """
    Read an SWC file and return a pandas DataFrame with columns:

        point_id
        section_type
        section_index
        parent_section_type
        parent_section_index
        x
        y
        z
        radius
        prev_point_id
    """

    if section_type_to_label is None:
        section_type_to_label = {
            0: "undefined",
            1: "soma",
            2: "axon",
            3: "basal_dendrite",
            4: "apical_dendrite",
            5: "apical_oblique",
            6: "apical_secondary_dendrite",
            7: "apical_secondary_oblique",
        }

    rows = []

    with open(swc_path, "r", encoding="utf-8", errors="ignore") as f:
        for line in f:
            line = line.strip()

            if not line or line.startswith("#"):
                continue

            parts = line.split()

            if len(parts) < 7:
                continue

            point_id = int(float(parts[0]))
            section_type_code = int(float(parts[1]))

            section_type = section_type_to_label.get(
                section_type_code,
                "undefined",
            )

            x = float(parts[2])
            y = float(parts[3])
            z = float(parts[4])
            radius = float(parts[5])
            prev_point_id = int(float(parts[6]))

            rows.append(
                [
                    point_id,
                    section_type,
                    x,
                    y,
                    z,
                    radius,
                    prev_point_id,
                ]
            )

    df = pd.DataFrame(
        rows,
        columns=[
            "point_id",
            "section_type",
            "x",
            "y",
            "z",
            "radius",
            "prev_point_id",
        ],
    )

    if df.empty:
        df["section_index"] = pd.Series(dtype=int)
   #     df["parent_section_type"] = pd.Series(dtype=str)
   #     df["parent_section_index"] = pd.Series(dtype=int)
   # else:
   #     df = add_section_index_by_type(df)
   #     df = add_parent_section_info(df)

   # df = df[
   #     [
   #         "point_id",
   #         "section_type",
   #         "section_index",
   #         "parent_section_type",
   #         "parent_section_index",
   #         "x",
   #         "y",
   #         "z",
   #         "radius",
   #         "prev_point_id",
   #     ]
   # ]

    #df.set_index(['section_type', 'section_index'], inplace=True)
    return df
```
